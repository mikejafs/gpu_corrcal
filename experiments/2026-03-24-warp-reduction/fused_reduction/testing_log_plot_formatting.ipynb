```python
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.ticker import ScalarFormatter
```

```python
xvals = np.arange(1, 11, 1)
yvals = np.exp(xvals)

print(xvals)
print(yvals)

fig, ax = plt.subplots()

ax.loglog(xvals, yvals, label = 'test')
ax.set_xticks(xvals)
# ax.set_xticklabels
ax.xaxis.set_major_formatter(ScalarFormatter())
ax.legend()
plt.show()
```
